```python
import pandas as pd

df = pd.read_csv("train.csv")
print(df.shape)
df.head()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)

print(df.isna().sum())          # missing values per column
print("Duplicates:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
total_revenue = df["Sales"].sum()
n_orders = df["Order ID"].nunique()
aov = total_revenue / n_orders

top_regions = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
top_products = df.groupby("Product Name")["Sales"].sum().nlargest(10)

print(f"Total revenue: ${total_revenue:,.2f}")
print(f"Number of orders: {n_orders:,}")
print(f"Average order value: ${aov:,.2f}")
print("\nRevenue by region:\n", top_regions)
print("\nTop 10 products:\n", top_products)
```

```python
monthly = df.set_index("Order Date")["Sales"].resample("ME").sum()
by_month = df.groupby(df["Order Date"].dt.month)["Sales"].sum()

print(by_month)
```

```python
import matplotlib.pyplot as plt
from google.colab import files

fig = plt.figure(figsize=(11.7, 8.3))
fig.suptitle("Sales Analysis Summary", fontsize=18, fontweight="bold")
fig.text(0.5, 0.91,
         f"Total revenue: ${total_revenue:,.0f}   |   Orders: {n_orders:,}   |   Avg order value: ${aov:,.2f}",
         ha="center", fontsize=12)

ax1 = fig.add_subplot(2, 2, 1)
monthly.plot(ax=ax1, color="#4c6ef5")
ax1.set_title("Monthly revenue")
ax1.set_xlabel("")

ax2 = fig.add_subplot(2, 2, 2)
tp = top_products.sort_values()
tp.index = tp.index.str[:28]
tp.plot.barh(ax=ax2, color="#7950f2")
ax2.set_title("Top 10 products by revenue")
ax2.set_ylabel("")

ax3 = fig.add_subplot(2, 2, 3)
top_regions.plot.bar(ax=ax3, color="#4c6ef5", rot=0)
ax3.set_title("Revenue by region")
ax3.set_xlabel("")

ax4 = fig.add_subplot(2, 2, 4)
by_month.plot.bar(ax=ax4, color="#7950f2", rot=0)
ax4.set_title("Seasonality: revenue by calendar month")
ax4.set_xlabel("Month")

plt.tight_layout(rect=[0, 0, 1, 0.88])
fig.savefig("sales_summary.pdf")
plt.show()
files.download("sales_summary.pdf")
```
